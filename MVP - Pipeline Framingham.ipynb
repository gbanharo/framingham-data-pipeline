## 4. Tratamento e Camada Silver

A camada Silver é responsável pela limpeza, padronização e preparação dos dados provenientes da camada Bronze.

Antes da aplicação das regras de tratamento, será realizada uma análise da qualidade dos dados para identificar valores ausentes, registros duplicados, tipos de dados inadequados e possíveis inconsistências. A partir desse diagnóstico serão definidas as transformações necessárias, buscando preservar a maior quantidade possível de informações relevantes para as análises e perguntas de negócio do projeto.

### 4.1 Diagnóstico da qualidade dos dados

Inicialmente, a tabela Bronze é utilizada como fonte para avaliar a qualidade do conjunto de dados. Nesta etapa serão investigados valores ausentes, duplicidades e os tipos de dados das variáveis, sem realizar alterações nos registros.

### 4.1.1 Valores ausentes

Antes da aplicação das regras de tratamento da camada Silver, foi realizada uma análise dos valores ausentes presentes na tabela Bronze.

A verificação considera tanto valores nulos quanto registros representados pelo texto `NA`, pois a inspeção inicial mostrou que algumas ausências da fonte foram carregadas dessa forma. O objetivo é identificar quais atributos necessitam de tratamento e apoiar a definição das transformações que serão aplicadas na camada Silver.

```python
from pyspark.sql import functions as F

# Leitura da camada Bronze
df_silver_base = spark.table("workspace.framingham_mvp.bronze_framingham")

# Contagem de valores ausentes (null ou NA)
expressoes_ausentes = []

for coluna in df_silver_base.columns:
    expressoes_ausentes.append(
        F.sum(
            F.when(
                F.col(coluna).isNull() |
                (F.upper(F.trim(F.col(coluna).cast("string"))) == "NA"),
                1
            ).otherwise(0)
        ).alias(coluna)
    )

df_ausentes = df_silver_base.select(expressoes_ausentes)

display(df_ausentes)
```

A análise identificou valores ausentes em sete atributos do conjunto de dados. A variável `glucose` apresentou a maior quantidade de ausências, com 388 registros, seguida por `education` (105), `BPMeds` (53), `totChol` (50), `cigsPerDay` (29), `BMI` (19) e `heartRate` (1). Os demais atributos não apresentaram valores ausentes.

Também foi observado que parte das ausências está representada pelo texto `NA` nos dados ingeridos, contribuindo para que algumas variáveis originalmente numéricas tenham sido inferidas pelo Spark como `string`. Esses casos serão tratados posteriormente durante a construção da camada Silver.

### 4.1.2 Registros duplicados

Além dos valores ausentes, foi verificada a existência de registros duplicados na camada Bronze. Essa análise permite identificar possíveis repetições antes da aplicação das regras de tratamento e construção da camada Silver.

```python
# Verificação de registros duplicados

# Quantidade total de registros
qtd_total = df_silver_base.count()

# Quantidade de registros distintos
qtd_distintos = df_silver_base.distinct().count()

# Diferença entre total e distintos
qtd_duplicados = qtd_total - qtd_distintos

print(f"Quantidade total de registros: {qtd_total}")
print(f"Quantidade de registros distintos: {qtd_distintos}")
print(f"Quantidade de registros duplicados: {qtd_duplicados}")
```

A verificação identificou 4.238 registros na tabela Bronze, sendo todos distintos. Portanto, não foram encontrados registros duplicados no conjunto de dados.

Esse resultado indica que não será necessária a remoção de duplicidades durante o tratamento da camada Silver, preservando-se todos os registros existentes em relação a esse critério de qualidade.

### 4.1.3 Tipos de dados e inconsistências

Além da análise de valores ausentes e duplicidades, foram verificados os tipos de dados atribuídos às variáveis durante a ingestão.

Essa verificação tem como objetivo identificar atributos cujo tipo inferido pelo Spark não corresponde à natureza esperada da informação, especialmente em colunas numéricas que contêm o valor textual `NA`. Essas inconsistências deverão ser corrigidas durante a etapa de tratamento e padronização da camada Silver.

```python
# Verificação dos tipos de dados da camada Bronze

tipos_dados = [
    (campo.name, campo.dataType.simpleString())
    for campo in df_silver_base.schema.fields
]

df_tipos = spark.createDataFrame(
    tipos_dados,
    ["atributo", "tipo_inferido"]
)

display(df_tipos)
```

A inspeção dos tipos de dados mostrou que algumas variáveis de natureza numérica foram inferidas pelo Spark como `string`: `education`, `cigsPerDay`, `BPMeds`, `totChol`, `BMI`, `heartRate` e `glucose`.

Esse comportamento está relacionado à presença do valor textual `NA` no arquivo de origem. Como a camada Bronze tem o objetivo de preservar os dados conforme recebidos, nenhuma correção foi realizada durante a ingestão.

Na camada Silver, esses valores serão convertidos para nulos e os atributos serão padronizados para tipos compatíveis com sua natureza, permitindo a realização adequada das análises posteriores.

## 4.2 Tratamento e padronização dos dados

Com base nos problemas identificados durante a análise de qualidade, são aplicadas as transformações necessárias para a construção da camada Silver.

Nesta etapa, os valores textuais `NA` são convertidos para valores nulos e os atributos são convertidos para tipos de dados compatíveis com sua natureza. A tabela Bronze permanece inalterada, sendo as transformações realizadas em um novo DataFrame destinado à camada Silver.

### 4.2.1 Conversão de `NA` e padronização dos tipos

```python
from pyspark.sql import functions as F

# Criação do DataFrame de trabalho da camada Silver
df_silver = df_silver_base

# Colunas que foram inferidas como string,
# mas representam informações numéricas
colunas_numericas_string = [
    "education",
    "cigsPerDay",
    "BPMeds",
    "totChol",
    "BMI",
    "heartRate",
    "glucose"
]

# Converte "NA" para null e padroniza as colunas como double
for coluna in colunas_numericas_string:
    df_silver = df_silver.withColumn(
        coluna,
        F.when(
            F.upper(F.trim(F.col(coluna))) == "NA",
            F.lit(None)
        )
        .otherwise(F.col(coluna))
        .cast("double")
    )

# Visualização do novo esquema
df_silver.printSchema()
```

A padronização dos tipos foi realizada com sucesso. Os valores textuais `NA` foram convertidos para valores nulos e as variáveis numéricas que haviam sido inferidas como `string` foram convertidas para o tipo `double`.

Com essa transformação, os atributos podem ser utilizados adequadamente em operações estatísticas e nas etapas posteriores de tratamento. A camada Bronze permanece inalterada, enquanto as modificações são aplicadas somente ao DataFrame de trabalho da camada Silver.

### 4.2.2 Análise para tratamento dos valores ausentes

Após a conversão dos valores `NA` para nulos, são analisadas as variáveis que apresentam dados ausentes. Essa etapa tem como objetivo definir uma estratégia de tratamento compatível com a natureza e a distribuição de cada atributo, evitando a aplicação indiscriminada de uma única regra de imputação.

```python
# Variáveis com valores ausentes
colunas_com_ausencias = [
    "education",
    "cigsPerDay",
    "BPMeds",
    "totChol",
    "BMI",
    "heartRate",
    "glucose"
]

# Estatísticas descritivas
display(
    df_silver.select(colunas_com_ausencias).summary(
        "count",
        "mean",
        "stddev",
        "min",
        "25%",
        "50%",
        "75%",
        "max"
    )
)
```

A análise das estatísticas descritivas foi utilizada para orientar a estratégia de tratamento dos valores ausentes. Para as variáveis numéricas contínuas (`cigsPerDay`, `totChol`, `BMI`, `heartRate` e `glucose`), será utilizada a mediana, por ser menos sensível à influência de valores extremos.

Para `education` e `BPMeds`, que representam informações discretas ou categóricas codificadas numericamente, será utilizada a moda. Dessa forma, o tratamento considera a natureza de cada atributo em vez de aplicar uma única regra de imputação a todo o conjunto de dados.

### 4.2.3 Tratamento dos valores ausentes

Com base na análise anterior, são aplicadas as estratégias de imputação definidas para cada grupo de variáveis. Os valores ausentes das variáveis numéricas contínuas são substituídos pela mediana de cada atributo, enquanto `education` e `BPMeds` são preenchidas pela respectiva moda.

As transformações são realizadas exclusivamente no DataFrame destinado à camada Silver, preservando os dados originais armazenados na camada Bronze.

```python
from pyspark.sql import functions as F

# Variáveis numéricas contínuas
# Tratamento dos valores ausentes pela mediana
colunas_mediana = [
    "cigsPerDay",
    "totChol",
    "BMI",
    "heartRate",
    "glucose"
]

for coluna in colunas_mediana:
    mediana = df_silver.approxQuantile(
        coluna,
        [0.5],
        0.001
    )[0]

    df_silver = df_silver.fillna({
        coluna: mediana
    })

# Variáveis discretas/categóricas
# Tratamento dos valores ausentes pela moda
colunas_moda = [
    "education",
    "BPMeds"
]

for coluna in colunas_moda:
    moda = (
        df_silver
        .where(F.col(coluna).isNotNull())
        .groupBy(coluna)
        .count()
        .orderBy(F.desc("count"))
        .first()[0]
    )

    df_silver = df_silver.fillna({
        coluna: moda
    })

print("Tratamento de valores ausentes concluído.")
```

#### Validação do tratamento

Após a imputação, é realizada uma nova verificação dos valores ausentes para confirmar a aplicação das regras definidas. Essa validação permite verificar se os atributos tratados deixaram de apresentar valores nulos antes da persistência dos dados na camada Silver.

```python
# Verificação dos valores ausentes após o tratamento

expressoes_validacao = [
    F.sum(
        F.when(
            F.col(coluna).isNull(),
            1
        ).otherwise(0)
    ).alias(coluna)
    for coluna in df_silver.columns
]

df_validacao_ausentes = df_silver.select(
    expressoes_validacao
)

display(df_validacao_ausentes)
```

A validação após a imputação confirmou que não permaneceram valores ausentes no DataFrame destinado à camada Silver. Todas as 16 variáveis apresentaram contagem igual a zero para valores nulos.

Dessa forma, os valores ausentes identificados na camada Bronze foram tratados conforme as regras definidas: mediana para as variáveis numéricas contínuas e moda para `education` e `BPMeds`. O conjunto de dados está, portanto, preparado para as próximas etapas de transformação e validação da camada Silver.

### 4.2.4 Validação de valores e domínios

Após o tratamento dos valores ausentes e a padronização dos tipos, são realizadas verificações de consistência sobre os valores presentes nas variáveis.

O objetivo é identificar possíveis valores fora dos domínios esperados, como códigos inválidos em atributos binários ou valores numéricos incompatíveis com a natureza das variáveis, antes da persistência definitiva da camada Silver.

```python
# Verificação dos valores distintos das variáveis categóricas/binárias

colunas_categoricas = [
    "male",
    "education",
    "currentSmoker",
    "BPMeds",
    "prevalentStroke",
    "prevalentHyp",
    "diabetes",
    "TenYearCHD"
]

for coluna in colunas_categoricas:
    print(f"\nValores distintos de {coluna}:")
    
    (
        df_silver
        .select(coluna)
        .distinct()
        .orderBy(coluna)
        .show()
    )
```

```python
# Verificação de valores mínimos e máximos das variáveis numéricas

colunas_numericas = [
    "age",
    "cigsPerDay",
    "totChol",
    "sysBP",
    "diaBP",
    "BMI",
    "heartRate",
    "glucose"
]

expressoes_limites = []

for coluna in colunas_numericas:
    expressoes_limites.append(
        F.min(coluna).alias(f"{coluna}_min")
    )
    expressoes_limites.append(
        F.max(coluna).alias(f"{coluna}_max")
    )

display(
    df_silver.select(expressoes_limites)
)
```

A verificação dos domínios permitiu avaliar a consistência dos valores após as etapas de padronização e tratamento. As variáveis binárias apresentaram somente os valores 0 e 1, enquanto `education` apresentou as categorias 1, 2, 3 e 4, não sendo identificados códigos inesperados nos atributos analisados.

Nas variáveis numéricas, foram observadas amplitudes distintas, incluindo valores elevados em atributos como `totChol`, `sysBP` e `glucose`. Entretanto, esses valores não foram considerados automaticamente como erros, uma vez que podem representar observações extremas válidas do conjunto de dados.

Dessa forma, optou-se por preservar esses registros na camada Silver, evitando exclusões ou alterações sem evidência suficiente de inconsistência.

### 4.2.5 Validação final do DataFrame Silver

Após a aplicação das regras de tratamento e padronização, é realizada uma validação final do DataFrame destinado à camada Silver.

Nesta etapa são verificadas a quantidade de registros e atributos, a existência de registros duplicados e a presença de valores ausentes. O objetivo é confirmar a integridade do conjunto de dados antes de sua persistência.

```python
# Validação final do DataFrame Silver

qtd_registros_silver = df_silver.count()
qtd_atributos_silver = len(df_silver.columns)

qtd_distintos_silver = df_silver.distinct().count()
qtd_duplicados_silver = qtd_registros_silver - qtd_distintos_silver

# Quantidade total de valores nulos
expressoes_nulos = [
    F.sum(
        F.when(F.col(coluna).isNull(), 1).otherwise(0)
    ).alias(coluna)
    for coluna in df_silver.columns
]

resultado_nulos = df_silver.select(expressoes_nulos).first()

total_nulos_silver = sum(resultado_nulos)

print(f"Quantidade de registros: {qtd_registros_silver}")
print(f"Quantidade de atributos: {qtd_atributos_silver}")
print(f"Quantidade de registros duplicados: {qtd_duplicados_silver}")
print(f"Quantidade total de valores ausentes: {total_nulos_silver}")
```

A validação final confirmou a integridade do DataFrame após as etapas de tratamento e padronização. O conjunto permaneceu com 4.238 registros e 16 atributos, sem registros duplicados e sem valores ausentes.

Dessa forma, o DataFrame encontra-se preparado para ser persistido na camada Silver, preservando todos os registros e atributos da camada Bronze e incorporando os tratamentos de valores ausentes e a padronização dos tipos de dados realizados nesta etapa.

## 4.3 Persistência da camada Silver

Após a conclusão das etapas de tratamento, padronização e validação, o DataFrame resultante é persistido na camada Silver em formato Delta.

A tabela é armazenada no Unity Catalog, no catálogo `workspace` e schema `framingham_mvp`, permitindo que os dados tratados sejam utilizados nas etapas posteriores do pipeline sem alterar a tabela original da camada Bronze.

```python
# Persistência dos dados na camada Silver

tabela_silver = "workspace.framingham_mvp.silver_framingham"

(
    df_silver.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(tabela_silver)
)

print(f"Tabela Silver criada: {tabela_silver}")
```

A camada Silver foi persistida com sucesso no Unity Catalog, em formato Delta, na tabela `workspace.framingham_mvp.silver_framingham`.

Essa tabela contém a versão tratada e padronizada dos dados provenientes da camada Bronze, com os tipos de dados corrigidos e os valores ausentes tratados, preservando os 4.238 registros e 16 atributos do conjunto original. A camada Bronze permanece inalterada, garantindo a rastreabilidade dos dados ao longo do pipeline.

## 4.4 Validação da camada Silver persistida

Após a persistência, a tabela Silver é carregada novamente a partir do Unity Catalog para verificar se os dados foram armazenados corretamente.

## 5.5 Relação das variáveis com a variável-alvo

Após a análise individual das variáveis, é avaliada sua relação com a variável-alvo `TenYearCHD`. O objetivo é identificar diferenças descritivas entre os indivíduos das classes 0 e 1, contribuindo para a compreensão das características associadas à ocorrência de doença coronariana no horizonte de dez anos.

Inicialmente, são comparadas as médias das principais variáveis numéricas entre as duas classes da variável-alvo. Essa análise possui caráter exploratório e não estabelece relações de causalidade, sendo utilizada para identificar padrões que poderão ser considerados nas etapas posteriores de modelagem.

```python
from pyspark.sql import functions as F

# Variáveis numéricas para comparação entre as classes
colunas_numericas = [
    "age",
    "cigsPerDay",
    "totChol",
    "sysBP",
    "diaBP",
    "BMI",
    "heartRate",
    "glucose"
]

# Média das variáveis numéricas por classe da variável-alvo
expressoes_media = [
    F.round(F.avg(coluna), 2).alias(f"media_{coluna}")
    for coluna in colunas_numericas
]

comparacao_target = (
    df_gold_base
    .groupBy("TenYearCHD")
    .agg(*expressoes_media)
    .orderBy("TenYearCHD")
)

display(comparacao_target)
```

Diferenças também foram observadas em `BMI`, com médias de 26,51 e 25,67, em `heartRate`, com 76,53 e 75,76, e em `glucose`, com 88,15 e 80,43, respectivamente para as classes 1 e 0.

De forma geral, todas as variáveis numéricas analisadas apresentaram médias superiores entre os registros pertencentes à classe 1 de `TenYearCHD`. As diferenças mais evidentes foram observadas em idade, pressão arterial sistólica e glicose. Esses resultados possuem caráter descritivo e não devem ser interpretados isoladamente como evidência de relação causal.

## 7.1 Ocorrência de doença coronariana por faixa etária

A primeira pergunta de negócio busca verificar como a ocorrência de doença coronariana em dez anos varia de acordo com a faixa etária dos pacientes.

Embora a seção 5.5 tenha mostrado que a idade média é maior entre os indivíduos com `TenYearCHD = 1`, nesta análise os pacientes são agrupados em faixas etárias para permitir uma comparação mais direta da frequência da classe positiva entre diferentes grupos de idade.

```python
from pyspark.sql import functions as F

# Criação das faixas etárias
df_faixa_etaria = (
    df_gold_validacao
    .withColumn(
        "faixa_etaria",
        F.when(F.col("age") < 40, "30-39")
         .when(F.col("age") < 50, "40-49")
         .when(F.col("age") < 60, "50-59")
         .otherwise("60-70")
    )
)

# Ocorrência de TenYearCHD por faixa etária
resultado_idade = (
    df_faixa_etaria
    .groupBy("faixa_etaria")
    .agg(
        F.count("*").alias("total_registros"),
        F.sum("TenYearCHD").alias("casos_CHD")
    )
    .withColumn(
        "percentual_CHD",
        F.round(
            (F.col("casos_CHD") / F.col("total_registros")) * 100,
            2
        )
    )
    .orderBy("faixa_etaria")
)

display(resultado_idade)
```

Os resultados mostram um aumento progressivo na ocorrência de `TenYearCHD = 1` conforme a faixa etária dos pacientes aumenta. Entre os indivíduos de 30 a 39 anos, 4,14% apresentaram ocorrência de doença coronariana no horizonte de dez anos. Esse percentual aumenta para 10,06% entre 40 e 49 anos, 19,73% entre 50 e 59 anos e atinge 27,68% entre 60 e 70 anos.

Dessa forma, no conjunto de dados analisado, observa-se uma associação entre faixas etárias mais elevadas e maior frequência de ocorrência de doença coronariana em dez anos. Esse resultado é consistente com a análise apresentada na seção 5.5, na qual os indivíduos com `TenYearCHD = 1` também apresentaram maior idade média.
